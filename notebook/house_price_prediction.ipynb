```python
house_data['per_sq_feet']=house_data['Price (USD)']/house_data['Square Feet']
house_data['per_sq_feet'].describe()
```

```python
house_data.info()
```

```python
X = house_data[['Bedrooms', 'Bathrooms', 'Square Feet', 'House Age', 'Location Rating']]
y = house_data['Price (USD)']
```

```python
print(X.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(
    X,y,
    test_size=0.3,
    random_state=30
)
```

```python
print(X_test.shape)
print(X_train.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
model1=LinearRegression()
model1.fit(X_train,y_train)
```

```python
y_pred=model1.predict(X_test)
y_pred[:5]
```

```python
print(y_pred[:5])
print(y_test.iloc[:5].values)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
mae=mean_absolute_error(y_pred,y_test)
mse=mean_squared_error(y_pred,y_test)
r2=r2_score(y_pred,y_test)
print("MAE:",mae)
print("MSE:",mse)
print("R2:",r2)
```

```python
train_pred = model1.predict(X_train)

print("Training R2:", r2_score(y_train, train_pred))
print("Testing R2:", r2_score(y_test, y_pred))
```

```python
print(model1.coef_)
```

```python
print(model1.intercept_)
```

```python
from sklearn.ensemble import RandomForestRegressor
model2=RandomForestRegressor(random_state=30)
model2.fit(X_train,y_train)
```

```python
y_pred2=model2.predict(X_test)
print(y_pred2[:5])
```

```python
print("MAE:", mean_absolute_error(y_test, y_pred2))
print("MSE:", mean_squared_error(y_test, y_pred2))
print("R2:", r2_score(y_test, y_pred2))
```

```python
train_pred2 = model2.predict(X_train)

print("Training R2:", r2_score(y_train, train_pred2))
print("Testing R2:", r2_score(y_test, y_pred2))
```
